```python
gender_summary = (
    df.groupby("gender")
    .agg(
        student_count=("gender", "count"),
        avg_math=("math score", "mean"),
        avg_reading=("reading score", "mean"),
        avg_writing=("writing score", "mean"),
        min_math=("math score", "min"),
        max_math=("math score", "max")
    )
    .reset_index()
)

gender_summary
```

This shows the average performance of all subjects for both genders

```python
test_prep_summary = (
    df.groupby("test preparation course")
    .agg(
        student_count=("test preparation course", "count"),
        avg_math=("math score", "mean"),
        avg_reading=("reading score", "mean"),
        avg_writing=("writing score", "mean")
    )
    .reset_index()
)

test_prep_summary
```

This shows if students perform better if they completed test preparation
